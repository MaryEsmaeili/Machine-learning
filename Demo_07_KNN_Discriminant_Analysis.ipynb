```
from matplotlib import pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs
from matplotlib.colors import ListedColormap

# Generate some toy data
X, y = make_blobs(n_samples=[10, 15, 25], n_features=2, centers=[[-1, -1], [0, 1], [1, 0]])
rgb_map = ListedColormap(('r', 'g', 'b'))

# Plot data
plt.grid(True)
plt.axhline(0.0, c='k')
plt.axvline(0.0, c='k')
plt.legend(*plt.scatter(X[:, 0], X[:, 1], c=y, cmap=rgb_map, edgecolors='w').legend_elements(), title='Classes:')
plt.title('$y(x)$')
plt.axis('square')
plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()
```

```
# k-Nearest Neighbors Implementation
class KNearestNeighbors:
    def __init__(self, n_neighbors=1):
        self.n_neighbors = n_neighbors
        self.classes = None
        self.X_train = None
        self.y_train_onehot = None

    def fit(self, X, y):
        self.classes = np.unique(y)
        self.X_train = X
        self.y_train_onehot = (y[:, np.newaxis] == self.classes)
        return self

    def predict_proba(self, X):
        sq_distance = np.sum((X[:, np.newaxis, :] - self.X_train[np.newaxis, :, :]) ** 2, axis=2)
        neighbors = np.argpartition(sq_distance, self.n_neighbors, axis=1)[:, :self.n_neighbors]
        y_proba = self.y_train_onehot[neighbors].mean(axis=1)
        return y_proba

    def predict(self, X):
        return self.classes[np.argmax(self.predict_proba(X), axis=1)]
```

```
# Test K-Nearest Neighbors
model = KNearestNeighbors(n_neighbors=3)
model.fit(X, y)
y_hat = model.predict(X)
y_hat
```

```
# Visualization of K-Nearest Neighbors
xx1, xx2 = np.meshgrid(np.linspace(-4.0, 4.0, 81), np.linspace(-4.0, 4.0, 81))
XX = np.array([xx1.ravel(), xx2.ravel()]).T
yy = model.predict_proba(XX).reshape(xx1.shape + (3, ))

plt.grid(True)
plt.axhline(0.0, c='k')
plt.axvline(0.0, c='k')
plt.imshow(yy, extent=(-4.0, 4.0, -4.0, 4.0), origin='lower', interpolation='bilinear', alpha=0.5)
plt.legend(*plt.scatter(X[:, 0], X[:, 1], c=y, cmap=rgb_map, edgecolors='w').legend_elements(), title='Classes:')
plt.title('$ŷ(x)$')
plt.axis('square')
plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()
```

```
# Linear Discriminant Analysis Implementation
class LinearDA:
    def __init__(self):
        self.classes = None

    def fit(self, X, y):
        self.classes, self.counts = np.unique(y, return_counts=True)
        self.means = {c: np.mean(X[y == c, :], axis=0) for c in self.classes}
        self.inv_cov = np.linalg.inv(np.cov([x_i - self.means[y_i] for x_i, y_i in zip(X, y)], ddof=0, rowvar=False))
        return self

    def predict_proba(self, X):
        y_proba = self.counts * np.exp([[-0.5 * (x_i - self.means[y_i]) @ self.inv_cov @ (x_i - self.means[y_i]) for y_i in self.classes] for x_i in X])
        y_proba = y_proba / np.sum(y_proba, axis=1, keepdims=True)
        return y_proba

    def predict(self, X):
        return self.classes[np.argmax(self.predict_proba(X), axis=1)]
```

```
# Test Linear Discriminant Analysis
model = LinearDA()
model.fit(X, y)
y_hat = model.predict(X)
y_hat
```

```
# Scikit-Learn Implementations
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
```
