```
from matplotlib import pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs

# Generate some toy-data
X, y = make_blobs(n_samples=[10, 20], n_features=2, centers=[[-1, -.5], [1, .5]])
X = np.hstack((X, np.ones((X.shape[0], 1))))

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdYlBu')
plt.title('$y(x)$')
plt.axis('square'); plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()
```

### Generate some toy-data

```
# Logistic regression
sigmoid = np.vectorize(lambda x:
    1.0 / (1.0 + np.exp(-x)) if x > 0 else np.exp(x) / (1.0 + np.exp(x))
)

xx = np.linspace(-6.0, 6.0, 61)

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.plot(xx, sigmoid(xx))
plt.title('$\sigma(x)$')
plt.show()

alpha = 10.0
costs = []
theta = np.zeros(3)
for _ in range(100):
    y_hat = sigmoid(X @ theta)
    error = y_hat - y
    theta -= alpha / y.size * X.T @ error
    cost = np.mean(-y * np.log(y_hat) - (1.0 - y) * np.log(1.0 - y_hat))
    costs.append(cost)

plt.grid(True)
plt.plot(costs, '-k')
plt.title("$J(t)$")
plt.yscale("log")
plt.show()

xx1, xx2 = np.meshgrid(np.linspace(-4.0, 4.0, 81), np.linspace(-4.0, 4.0, 81))
XX = np.array([xx1.ravel(), xx2.ravel(), np.ones(xx1.size)]).T
yy = sigmoid(XX @ theta).reshape(xx1.shape)

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.imshow(yy, extent=(-4.0, 4.0, -4.0, 4.0), origin='lower', interpolation='bilinear', cmap='RdYlBu', vmin=0.0, vmax=1.0, alpha=0.5)
plt.contour(xx1, xx2, yy, [0.5], linewidths=0.5, linestyles='dashed')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdYlBu')
plt.colorbar()
plt.title('$\hat{y}(x)$')
plt.axis('square'); plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()

cost = np.mean(-y * np.log(y_hat) - (1.0 - y) * np.log(1.0 - y_hat))
cost
```

### Logistic regression

```
# Scikit-learn
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(penalty=None)
model.fit(X[:, :-1], y)

model.coef_, model.intercept_
model = LogisticRegression(fit_intercept=False, penalty=None)
model.fit(X, y)

y_hat = model.predict_proba(X)[:, 1]

xx1, xx2 = np.meshgrid(np.linspace(-4.0, 4.0, 81), np.linspace(-4.0, 4.0, 81))
XX = np.array([xx1.ravel(), xx2.ravel(), np.ones(xx1.size)]).T
yy = model.predict_proba(XX)[:, 1].reshape(xx1.shape)

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.imshow(yy, extent=(-4.0, 4.0, -4.0, 4.0), origin='lower', interpolation='bilinear', cmap='RdYlBu', vmin=0.0, vmax=1.0, alpha=0.5)
plt.contour(xx1, xx2, yy, [0.5], linewidths=0.5, linestyles='dashed')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdYlBu')
plt.colorbar()
plt.title('$\hat{y}(x)$')
plt.axis('square'); plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()

cost = np.mean(-y * np.log(y_hat) - (1.0 - y) * np.log(1.0 - y_hat))
cost
```
