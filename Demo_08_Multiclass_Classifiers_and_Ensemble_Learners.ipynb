```

from matplotlib import pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs

# Generate some toy data
X, y = make_blobs(n_samples=[10, 15, 25], n_features=2, centers=[[-1, -1], [0, 1], [1, 0]])

X, y
```

```

from matplotlib.colors import ListedColormap
rgb_map = ListedColormap(('r', 'g', 'b'))

plt.grid(True)
plt.axhline(0.0, c='k')
plt.axvline(0.0, c='k')
plt.legend(*plt.scatter(X[:, 0], X[:, 1], c=y, cmap=rgb_map, edgecolors='w').legend_elements(), title='Classes:')
plt.title('$y(x)$')
plt.axis('square')
plt.axis((-4.0, 4.0, -4.0, 4.0))
plt.show()
```

```

from sklearn.base import clone

class MulticlassOVR():
    def __init__(self, classifier):
        self.classifier = classifier
        self.classifiers = None
        self.classes = None
        
    def __repr__(self):
        return f'MulticlassOVR({self.classifier})'
        
    def fit(self, X, y):
        self.classifiers = dict()
        self.classes = np.unique(y)
        for c in self.classes:
            subset = (y == c)
            self.classifiers[c] = clone(self.classifier)
            self.classifiers[c].fit(X, subset)
        return self
            
    def predict_proba(self, X):
        y_proba = np.zeros((X.shape[0], len(self.classes)))
        for i, c in enumerate(self.classes):
            y_proba[:, i] = self.classifiers[c].predict_proba(X)[:, 1]
        y_proba = y_proba / np.sum(y_proba, axis=1, keepdims=True)
        return y_proba
            
    def predict(self, X):
        return self.classes[np.argmax(self.predict_proba(X), axis=1)]
from sklearn.linear_model import LogisticRegression
model = MulticlassOVR(LogisticRegression())

model
```
