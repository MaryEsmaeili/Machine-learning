## Linear regression: matrix method

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-1.6,1.6,17)
```

```python
y = np.cos(x)
```

```python
plt.plot(x,y,'o')
plt.show()
```

```python
# fit with a parabola: y = theta0+ theta1 * x +theta2*x^2
X = np.array([np.ones_like(x),x,x**2]).T
X
```

```python
theta = np.linalg.inv(X.T @ X) @ (X.T @ y)
theta
```

```python
# theta, *_ = np.linalg.solve(X.T @ X, X.T @y)
# theta
```

```python
# theta = np.linalg.lstsq(X, y, rcond=None)
# theta
```

```python
# yhat = theta[0] + theta[1] * x + theta[2] * x**2
# imporoved version of yhat
yhat= X @ theta
plt.plot(x,y,'o')
plt.plot(x,yhat,'.')
plt.show()
```

```python
error = yhat -y 
error
```

```python
cost = 0.5 / len(y) * error.T @ error
cost
```

## linear regression : gradient descent method

```python
# we need an initial value to start with
```

```python
# gradient descent
theta = np.zeros(X.shape[1])
costs = []
#learning rate
alpha = 5e-1
for _ in range(100):
    yhat = X @ theta
    error = yhat -y
    theta =  theta - alpha/len(y)*X.T @ error
    cost =  0.5 / len(y) * error.T @ error
    costs.append(cost)
theta
```

```python
# plt.plot(costs - costs[-1])
plt.plot(costs)
plt.yscale("log")
plt.show()
```
