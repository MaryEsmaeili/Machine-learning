## Linear regression: regularisation
Note: This notebook is a 'cleaned-up' version of the code written in class; it contains only minimal textual explanation. Please refer to the lessons for further information about what is done, and why.

```python
from matplotlib import pyplot as plt
import numpy as np

# Generate some toy-data
x = np.linspace(-1.0, 1.0, 11)
y = x + np.random.randn(*x.shape)

x, y
plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.plot(x, y, 'o')
plt.title('$y(x)$')
plt.show()
```

### Generate some toy-data

```python
# Linear regression
X = np.vander(x, 9)
theta = np.linalg.solve(X.T @ X, X.T @ y)
y_hat = X @ theta

xx = np.linspace(-1.0, 1.0, 101)

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.plot(x, y, 'o')
plt.plot(x, y_hat, ':.')
plt.plot(xx, np.polyval(theta, xx), lw=0.5)
plt.title('$y(x)$')
plt.show()

error = y_hat - y
cost = 0.5 * np.mean(error ** 2)
cost
```

### Linear regression

```python
# Ridge regression
lambda2 = 1.0
theta = np.linalg.solve(X.T @ X + lambda2 * np.eye(X.shape[1]), X.T @ y)
y_hat = X @ theta

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.plot(x, y, 'o')
plt.plot(x, y_hat, ':.')
plt.plot(xx, np.polyval(theta, xx), lw=0.5)
plt.title('$y(x)$')
plt.show()

error = y_hat - y
cost = 0.5 / y.size * (np.sum(error ** 2) + lambda2 * np.sum(theta ** 2))
cost
```

### Ridge regression

```python
# Gradient descent
alpha = 0.5
lambda2 = 1.0
costs = []
theta = np.zeros(X.shape[1])
for _ in range(100):
    y_hat = X @ theta
    error = y_hat - y
    theta -= alpha / y.size * (X.T @ error + lambda2 * theta)
    cost = 0.5 / y.size * (np.sum(error ** 2) + lambda2 * np.sum(theta ** 2))
    costs.append(cost)

plt.grid(True)
plt.plot(costs, '-k')
plt.title("$J(t)$")
plt.yscale("log")
plt.show()
```

### Gradient descent

```python
# Scikit-learn
from sklearn.linear_model import Ridge
model = Ridge(alpha=lambda2)
model.fit(X[:, :-1], y)

model.coef_, model.intercept_
model = Ridge(fit_intercept=False, alpha=lambda2)
model.fit(X, y)

y_hat = model.predict(X)

plt.grid(True); plt.axhline(0.0, c='k'); plt.axvline(0.0, c='k')
plt.plot(x, y, 'o')
plt.plot(x, y_hat, ':.')
plt.plot(xx, np.polyval(theta, xx), lw=0.5)
plt.title('$y(x)$')
plt.show()

error = y_hat - y
cost = 0.5 / y.size * (np.sum(error ** 2) + lambda2 * np.sum(theta ** 2))
cost
```

### Concepts to Understand Better:
1. **Linear Regression**:
   - The mathematical formulation: \( \hat{Y} = X \Theta \), where \(X\) is the feature matrix and \(\Theta\) represents the coefficients.
   - Matrix inversion approach to solving linear regression: \( \Theta = (X^T X)^{-1} X^T y \).

2. **Gradient Descent**:
   - The cost function and how gradient descent minimizes it.
   - The update rule for the weights in gradient descent: 
     \[ \Theta := \Theta - \alpha \frac{\partial J(\Theta)}{\partial \Theta} \]
   - Learning rate (\(\alpha\)) and how it affects convergence.

3. **Overfitting**:
   - What it means in the context of polynomial regression.
   - High-degree polynomial models can lead to high variance (overfitting), where the model fits the training data too closely and performs poorly on unseen data.

4. **Regularization**:
   - **Ridge Regression (L2 Regularization)**: Penalizes the sum of squared coefficients.
   - **Lasso Regression (L1 Regularization)**: Penalizes the sum of absolute coefficients.
   - These techniques help control overfitting by discouraging large coefficients and making the model more generalizable.
